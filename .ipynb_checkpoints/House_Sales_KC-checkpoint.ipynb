```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# Importing dataset
housing_data = pd.read_csv('kc_house_data.csv')
```

```python
housing_data
```

```python
# Cleaning of data
housing_data = housing_data.drop(['id','sqft_living15', 'sqft_lot15', 'date'], axis = 1)
housing_data
```

```python
housing_data.info()
```

```python
housing_data.describe()
```

## First we will make multiple correlation matrix and graphs to understand the realtionship and our data
1. `first we'll multiple graph of X with relation to y (bedrooms, sqft_living, floors, sqft_lot, yr_built,.....)`
2. `We will check columns and try to understand them`

**Step : 1**

```python
# creating custom colums to understand our data  
df = housing_data.drop(['waterfront','view', 'lat', 'long','zipcode'], axis = 1)
df.head()
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.bedrooms, y = df.price, c = 'salmon')
plt.xlabel('Rooms')
plt.ylabel('Price');
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.bathrooms, y = df.price, c = 'blue')
plt.xlabel('Bathrooms')
plt.ylabel('Price');
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.sqft_living, y = df.price)
plt.xlabel('Living Area ')
plt.ylabel('Price');
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.sqft_lot, y = df.price)
plt.xlabel('Total Area')
plt.ylabel('Price');
```

```python
plt.figure(figsize = (10,6))
plt.scatter(df.sqft_lot, df.price, label = 'Total Area', alpha = 0.6)
plt.scatter(df.sqft_living, df.price, label = 'Living Area', alpha = 0.6)
plt.xlabel('Area')
plt.ylabel('Price')
plt.legend()
plt.title('Price vs Total Area and Living Area')
plt.show()
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.condition, y = df.price)
plt.xlabel('Condition')
plt.ylabel('Price');
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.grade, y = df.price)
plt.xlabel('Grade')
plt.ylabel('Price');
```

```python
plt.figure(figsize = (10,6))
plt.scatter(x = df.yr_built, y = df.price)
plt.xlabel('yr_built')
plt.ylabel('Price');
```
